```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt=ChatPromptTemplate.from_messages(
    [('system','Your are a Doctor. Do not answer anything not related to the medical domain'),
     MessagesPlaceholder(variable_name="chat_history"),
     ('human','{input}')]
)
```

```python
GROQ_API_KEY = "YOUR_GROQ_API_KEY"  # Replace with your actual Groq API key

grok_model = ChatGroq(model="llama-3.3-70b-versatile", api_key=GROQ_API_KEY)
```

```python
memory=ConversationBufferMemory(
    memory_key='chat_history',
    return_messages=True
)
```

```python
pipeline=ConversationChain(
    llm=grok_model,
    verbose=False,
    prompt=prompt,
    memory=memory
)
```

```python
while True:
    user_input = input("User: ").strip()
    if not user_input:
        continue

    if user_input.lower() == "exit":
        clear_choice = input("Clear memory before exit? (y/n): ").strip().lower()
        if clear_choice == "y":
            memory.clear()
            print("Conversation memory cleared.")
        else:
            print("Conversation memory kept.")
        break

    response = pipeline.invoke(input=user_input)
    answer = response["response"]
    print(f"Doctor: {answer}", flush=True)
```
