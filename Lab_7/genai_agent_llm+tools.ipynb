### Custom Tools

```python
def sum1(a,b):
  return a+b
```

```python
def sum1(a,b):
  """ this function is doing the sum of two numbers"""
  return a+b
```

```python
from langchain_core.tools import tool
```

```python
@tool
def sum1(a:int,b:int)->int:
  """ this function is doing the sum of two numbers"""
  return a+b
```

```python
sum1.description
```

```python
sum1.args
```

```python
result=sum1.invoke({"a":5,"b":6})
```

```python
result
```

```python
@tool
def mul(a:int,b:int)->int:
  "Multiplication of two numbers"
  return a*b
```

```python
mul.description
```

```python
mul.args
```

```python
result2=mul.invoke({"a":5,"b":6})
```

```python
result2
```

```python
from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper
```

```python
# 1. Configure the API wrapper to limit your results
api_wrapper = WikipediaAPIWrapper(top_k_results=2, doc_content_chars_max=1000)

# 2. Initialize the Wikipedia tool
wikipedia_tool = WikipediaQueryRun(api_wrapper=api_wrapper)

# 3. Run a query directly
response = wikipedia_tool.run({"query": "Generative Artificial Intelligence"})
print(response)
```

```python
wikipedia_tool.description
```

```python
wikipedia_tool.args
```

```python
!pip install langchain-groq
from langchain_groq import ChatGroq
```

```python
GROQ_API_KEY = "YOUR_GROQ_API_KEY"  # Replace with your actual API key

llm = ChatGroq( 
    model="openai/gpt-oss-120b",
    api_key=GROQ_API_KEY,
)
```

```python
llm.invoke("helulu")
```

```python
tools=[sum1,mul,wikipedia_tool,ddg_tool]
```

```python
llm_with_tools=llm.bind_tools(tools)
```

```python
response=llm_with_tools.invoke("tell me current news about ai 2026??")
```

```python
response
```

```python
from langchain.agents import create_agent
```

```python
agent=create_agent(
    model=llm,
    tools=tools
)
```

```python
agent.invoke({
        "messages":[{
                "role":"user",
                "content":"What are the top AI news?"
            }
        ]  
    }
)
```
