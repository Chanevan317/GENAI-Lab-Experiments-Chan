```python
from langchain_community.document_loaders import PyMuPDFLoader
```

```python
!pip install pypdf pymupdf
```

```python
text=PyMuPDFLoader("/content/FIFA World Cup 2026.pdf")
```

```python
text_load=text.load()
```

```python
text_load
```

```python
from langchain_text_splitters  import CharacterTextSplitter
```

```python
splitter=CharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=50,
)
```

```python
chunks=splitter.split_documents(text_load)
```

```python
len(chunks)
```

```python
for i in chunks:
  print(i.page_content)
  print("--------------------------------------------------------")
```
