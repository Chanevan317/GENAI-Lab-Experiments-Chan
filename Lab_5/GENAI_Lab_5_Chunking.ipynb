```python
from langchain_community.document_loaders import PyMuPDFLoader
```

```python
text=PyMuPDFLoader("/content/RAG_paper.pdf")
```

```python
text_load=text.load()
```

```python
text_load
```

```python
from langchain_text_splitters  import CharacterTextSplitter
```

```python
splitter1=CharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=50,
)
```

```python
chunks1=splitter.split_documents(text_load)
```

```python
len(chunks1)
```

```python
for i in chunks1:
  print(i.page_content)
  print("--------------------------------------------------------")
```
