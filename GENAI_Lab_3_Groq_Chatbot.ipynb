```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt=ChatPromptTemplate.from_messages(
    [('system','Your are a Doctor. Do not answer anything not related to the medical domain'),
     MessagesPlaceholder(variable_name="chat_history"),
     ('human','{input}')]
)
```

```python
import os

GROQ_API_KEY = os.getenv("GROQ_API_KEY", "YOUR_GROQ_API_KEY")

grok_model = ChatGroq(model="llama-3.3-70b-versatile", api_key=GROQ_API_KEY)
```

```python
memory=ConversationBufferMemory(
    memory_key='chat_history',
    return_messages=True
)
```

```python
pipeline=ConversationChain(
    llm=grok_model,
    verbose=True,
    prompt=prompt,
    memory=memory
)
```

```python
user_input=input('Enter your question: ')

print(user_input)
```

```python
while True:
  user_input=input("User: ")
  if user_input=='exit':
    break
  response=pipeline.invoke(input=user_input)
  print(response)
```
